```python
import pandas as pd
import numpy as np
import re
import joblib
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import TfidfVectorizer

print("Loading data...")
train_df = pd.read_csv('../data/raw/train.csv')
val_df = pd.read_csv('../data/raw/dev.csv')
test_df = pd.read_csv('../data/raw/test.csv')

print("Combining text...")
train_texts = (train_df['article'] + ' ' + train_df['question']).tolist()
val_texts = (val_df['article'] + ' ' + val_df['question']).tolist()
test_texts = (test_df['article'] + ' ' + test_df['question']).tolist()

print("TF-IDF vectorizing...")
tfidf = TfidfVectorizer(max_features=500, stop_words='english', sublinear_tf=True)
X_train = tfidf.fit_transform(train_texts)
X_val = tfidf.transform(val_texts)
X_test = tfidf.transform(test_texts)

print("Encoding labels...")
le = LabelEncoder()
y_train = le.fit_transform(train_df['answer'])
y_val = le.transform(val_df['answer'])
y_test = le.transform(test_df['answer'])

print("Saving...")
import os
os.makedirs('../data/processed', exist_ok=True)
joblib.dump(tfidf, '../data/processed/tfidf_vectorizer.pkl')
joblib.dump(le, '../data/processed/label_encoder.pkl')
joblib.dump(train_df, '../data/processed/train_df.pkl')
joblib.dump(val_df, '../data/processed/val_df.pkl')
joblib.dump(test_df, '../data/processed/test_df.pkl')

print("Shapes:")
print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("X_test:", X_test.shape)
print("Done!")
```
